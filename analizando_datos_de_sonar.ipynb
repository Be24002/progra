# Definición del problema


El enfoque de este proyecto será el conjunto de datos Sonar Mines vs Rocks. El problema consiste en predecir si un objeto es de metal o una roca a partir de los datos de retorno del sonar. Cada patrón es un conjunto de 60 números en el rango de 0.0 a 1.0. Cada número representa la energía dentro de una banda de frecuencia particular, integrada en un cierto período de tiempo. La etiqueta asociada a cada registro contiene la letra R si el objeto es una roca y M si es una mina (cilindro de metal). Los números en las etiquetas están en orden creciente según el ángulo de aspecto, pero no codifican directamente dicho ángulo.

# Cargando librerias

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
sns.set_style("whitegrid")
```

# Cargando los datos

```python
# Load the dataset
df = pd.read_csv("sonar_all_data.csv", header = None)
# Rename columns using a dictionary
df = df.rename(columns={i: f'var{i+1}' for i in range(df.shape[1])})
```

En la celda anterior hemos cargado los datos usando la libreria pandas.
